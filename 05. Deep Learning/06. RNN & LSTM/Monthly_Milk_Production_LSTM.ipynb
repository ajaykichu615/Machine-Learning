```python
df.dropna(inplace=True)
```

```python
from sklearn.preprocessing import MinMaxScaler
minmax=MinMaxScaler()
X_scaled=minmax.fit_transform(df)
```

```python
X_scaled.shape
```

```python
def train_split(data, train_size=0.7):
  rows,column=data.shape
  train_number=round(rows*train_size)
  train_data=data[:train_number]
  test_data=data[train_number:]
  return train_data,test_data
```

```python
train,test=train_split(X_scaled, train_size=0.8)
```

```python
df.head(15)
```

```python
from tensorflow.keras.preprocessing.sequence import TimeseriesGenerator
train_generator=TimeseriesGenerator(train,train,length=3,batch_size=1)
```

```python
from tensorflow.keras.layers import Dense,LSTM,Input
from tensorflow.keras.models import Sequential
```

```python
model=Sequential([Input(shape=(3,1)),
                 LSTM(100,return_sequences=True,activation='relu'),
                 LSTM(50,return_sequences=True,activation='relu'),
                 LSTM(25,return_sequences=True,activation='relu'),
                 Dense(1)])
```

```python
model.compile(optimizer='adam',loss='mse')
```

```python
model.fit(train_generator,epochs=20)
```
