```python
vectordb={}
known_images=[('Random dude','face.jpg'),('Mammootty','Mammootty,_2022.jpg'),('Rajuvettan','Raju.jpg')]
for i,j in known_images:
    img_array=face_recognition.load_image_file(j)
    face_encodings=face_recognition.face_encodings(img_array)
    if face_encodings:
        face_encodings=face_encodings[0]
        vectordb[i]=face_encodings
```

```python
len(vectordb)
```

```python
unknown_array=face_recognition.load_image_file('Ajay.jpg')
unknown_vectors=face_recognition.face_encodings(unknown_array)
if unknown_vectors:
    unknown_vector = unknown_vectors[0]
    ret=face_recognition.compare_faces(unknown_vector,list(vectordb.values()),tolerance=0.5)
    if True in ret:
        ind=ret.index(True)
        name=list(vectordb.keys())[ind]
        print(name)
    else:
        print('No match found')    
```
