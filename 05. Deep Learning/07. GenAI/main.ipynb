```python
import google.generativeai as genai
```

```python
import api
genai.configure(api_key=api.api_key)
```

```python
for model in genai.list_models():
    print(model.name)
```

```python
model = genai.GenerativeModel('models/gemini-1.5-flash')
```

```python
model.generate_content('What is Data Science?').text
```

```python
chat = model.start_chat(history=[])
print(chat.send_message('Hello what is machine learning?').text)
```

```python
print(chat.send_message('What are its applicaitons?'))
```

```python
chat = model.start_chat(history=[]) # clears history
```

```python
print(chat.send_message('What are its applicaitons?').text) # history reset, model unable to answer to the context
```
