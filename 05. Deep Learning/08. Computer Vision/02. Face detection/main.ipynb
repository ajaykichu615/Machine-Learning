```python
import cv2
```

```python
face_cascade=cv2.CascadeClassifier("haarcascade_frontalface_default.xml")
```

```python
face_cascade
```

```python
img_array=cv2.imread('face.jpg')
```

```python
img_array
```

```python
img_array_grey=cv2.cvtColor(img_array,cv2.COLOR_BGR2GRAY)
res=face_cascade.detectMultiScale(img_array_grey,minNeighbors=5)
```

```python
(x1,y1)=187,49
x2=x1+153
y2=y1+153
cv2.rectangle(img_array,(x1,y1),(x2,y2),(0,255,255),2)
cv2.imshow('face', img_array)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```

```python
# Trying face detection from camera

cap=cv2.VideoCapture(0)
while True:
    ret,frame=cap.read()
    if not ret:
        print("no Frame")
        break
    height,width,_=frame.shape
    frame_gray=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    result=face_cascade.detectMultiScale(frame_gray,minNeighbors=3)
    if len(result)!=0:
        for i in result:
            x1,y1,w,h=i
            x2,y2=x1+w,y1+h
            cv2.rectangle(frame,(x1,y1),(x2,y2),(255,250,250),1)
    else:
        cv2.putText(frame,"No face detected",(20,height-20),6,1,(255,255,255),1)
    cv2.imshow("face detection",frame)   
    if cv2.waitKey(16)==ord("k"):
        cv2.destroyAllWindows()
        break
cap.release()
```
