```python
import cv2
```

```python
img_array=cv2.imread("pexels-jvdm-1599791.jpg")
```

```python
if img_array is not None:
	print(img_array.shape)
else:
	print("Error: Image not loaded. Please check the file path.")
```

```python
height, width, _ = img_array.shape
```

```python
height=600
width=500
img_resize=cv2.resize(img_array,(width,height))  # changing the resolution of image
```

```python
img_resize.shape
```

```python
cv2.imshow("a", img_resize)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.imshow(img_resize)
```

```python
img_resize.shape
```

```python
cv2.line(img_resize,(0,0),(500,600),(0,255,0),3) # draws a line across the image at the specified coordinates. 3 is for thickness of the line.
```

```python
cv2.imshow("line", img_resize)
cv2.waitKey(1000)
cv2.destroyAllWindows()
```

```python
cv2.circle(img_resize,(width//2,height//2),200,(32,111,44),2)
cv2.imshow("circle", img_resize)
cv2.waitKey(10000)
cv2.destroyAllWindows()
```

```python
cv2.rectangle(img_resize,(width//2,0),(width-4,height//2),(23,32,12),3)
cv2.imshow("rectangle", img_resize)
cv2.waitKey(6000) # milliseconds
cv2.destroyAllWindows()
```

```python
cv2.putText(img_resize,"I am Iron man",(5,height-5),cv2.FONT_HERSHEY_COMPLEX,1,(255,255,255),2)
cv2.imshow("text", img_resize)
cv2.waitKey(6000)
cv2.destroyAllWindows()
```

```python
cap=cv2.VideoCapture(0) # Works if you have camera device attached. 0 is the default camera.
cap.read()
```

```python
ret,frame=cap.read()
if ret:
    cv2.imshow('frame', frame)
    cv2.waitKey(100)
    cv2.destroyAllWindows()
else:
    print('No device attached or there is a problem with the device')

cap.release() # to release the device access
```

```python
while True:
    ret,frame=cap.read()
    if not ret:
        print("No frame")
        break
    frame=cv2.flip(frame,1) # to get mirror image
    h,w,_=frame.shape
    cv2.putText(frame,"I am Iron man", (width-100,40),6,1,(255,255,255),2)
    cv2.imshow('Frame',frame)
    if cv2.waitKey(20)==ord("e"):
        print('Video stopped')
        cv2.destroyAllWindows()
        cap.release()
        break
```

```python
img_res_rgb=cv2.cvtColor(img_resize,cv2.COLOR_BGR2RGB) # converting BGR in cv2 to RGB widely used especially in matplot
plt.imshow(img_res_rgb)
```
